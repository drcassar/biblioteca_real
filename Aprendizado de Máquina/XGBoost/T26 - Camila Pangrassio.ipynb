## Introdução:

XGBoost é um **algoritmo de aprendizado de máquina** supervisionado muito utilizado para resolver problemas de previsão de dados em tabelas. Ele é conhecido pela sua eficiência, velocidade e grande capacidade de escalar grandes conjuntos de dados. Por isso, se tornou uma opção popular em competições de ciência de dados e em projetos para o ramo industrial, graças à sua capacidade de gerar resultados precisos de forma eficiente. 

Esse algoritmo é baseado em **árvores de decisão que utiliza a técnica de gradient boosting**, ou seja, é uma versão otimizada do método de gradient boosting. A ideia central é usar modelos fracos sequencialmente, corrigindo os erros dos anteriores, o que melhora a acurácia as previsões seguintes.

Inicialmente, o processo começa com um modelo inicial simples, como uma árvore de decisão pouco profunda, que não é muito preciso. Depois disso, este modelo inicial é usado para fazer previsões nos dados de treinamento, em que a diferença entre essas previsões e os valores reais é chamada de erro ou resíduo. Um novo modelo é treinado levando em consideração os erros do modelo anterior, focando em corrigi-los. Ao realizar esse processo diversas vezes, é criado uma sequência de modelos que se complementam. No final desse processo, com a combinação ponderada de todos os modelos fracos, temos um modelo altamente mais preciso.

## Boosting

Para iniciar a ánalise, vamos entender o conceito de boosting, que é um método de **aprendizado em conjunto** que combina diversos aprendizes fracos e os transforma em um aprendiz forte, visando minimizar os erros de treinamento. No boosting, uma amostra aleatória de dados é selecionada, ajustada a um modelo e, em seguida, treinada para modelos seguintes, tentando criar modelos novos que compensam os pontos fracos de modelos anteriores.

Esse aprendizado coletivo fundamenta-se na ideia de que decisões tomadas em grupos são melhores que decisões feitas individualmente, dado que um único modelo esta sujeito a alta variância ou alto viés (ou seja, overfitting), e por isso pode apresentar um desempenho fraco. Entretanto, a combinação de diversos modelos reduz o overfitting, gerando melhores resultados.

Esse aprendizado em conjunto é ilustrado pela a imagem a seguir, com duas árvores que utilizam pesos diferentes para se complementarem em uma previsão.

## Gradient Boosting

Depois do Boosting, tivemos a criação de variações mais avançadas desse algoritmo.

A primeira delas foi o **AdaBoosting ou Boosting adaptativo**, em que o modelo opera identificando pontos de dados classificados incorretamente e ajustando seus pesos para minimizar o erro de treinamento. O modelo continua otimizado em sequência até produzir a previsão mais forte. 

**Gradiente descendente**: é um algoritmo de otimização que treina modelos de aprendizado de máquina ao minimizar os erros entre os resultados previstos e os reais. A função de custo dentro da descida do gradiente avalia a precisão a cada iteração de atualização de parâmetros. Até que esta função esteja próxima ou igual a zero, o modelo continuará ajustando seus parâmetros para obter o menor erro possível. 

A partir de um ponto inicial, encontramos a derivada e, com isso, podemos usar uma linha tangente para observar a inclinação da curva. A inclinação informará as atualizações dos pesos e o viés (parâmetros do modelo). A inclinação no ponto de partida será mais acentuada, mas à medida que novos parâmetros são gerados, a inclinação deve gradualmente diminuir até atingir o ponto mais baixo da curva, conhecido como ponto de convergência (ou minimo local), sendo este o lugar em que ocorre o menor erro.

Em seguida, foi desenvolvido o **gradient boosting**, que funciona adicionando sequencialmente preditores a um conjunto, com cada um corrigindo os erros de seu antecessor. No entanto, em vez de alterar os pesos dos pontos de dados como o AdaBoost, o gradient boosting treina nos erros residuais (diferença entre os valores reais e os valores previstos pelo modelo vigente) da previsão anterior. O nome do método gradient boosting foi dado pois combina o algoritmo de gradiente descendente e o boosting.

## eXtreme Gradient Boosting

Ao evoluir mais um passo do método anterior, temos o XGBoost, que é uma implementação do gradient boosting projetada para velocidade e escala computacional. O XGBoost aproveita vários núcleos na CPU, possibilitando que o aprendizado ocorra em paralelo durante o treinamento. Além disso, ele é conhecido por sua **eficiência na otimização da função de perda**, utilizando as informações da matriz Hessiana, o que o torna mais eficiente no treinamento em comparação com o gradient boosting 

O XGBoost incorpora técnicas de regularização, como a regularização L1 (Regressão Lasso) e a regularização L2 (Regressão Ridge), para evitar o sobreajuste e melhorar a generalização do modelo, ajudando a impedir que os modelos se tornem muito complexos e se ajustem demais aos dados. A regularização L1 simplifica o modelo removendo características menos importantes, enquanto a regularização L2 mantém o modelo estável equilibrando a influência de diferentes características.

Este algoritmo oferece flexibilidade em termos de personalização e **ajuste de hiperparâmetros**, permitindo que os usuários ajustem o modelo de acordo com casos de uso e objetivos específicos. Assim, é visto que o ajuste dos hiperparâmetros é uma das funcionalidades que amplia o sucesso do método e o torna tão poderoso.


### Passo a passo do processo do XGBoost

**Modelos Fracos:** O processo começa com a criação de um modelo inicial simples, como uma árvore de decisão com pouca profundidade. Este modelo inicial pode não ser muito preciso.

**Cálculo dos Resíduos:** O modelo inicial é usado para fazer previsões nos dados de treinamento. A diferença entre as previsões e os valores reais é chamada de resíduo ou erro.

**Novo Modelo nos Resíduos:** Um novo modelo é treinado para prever os resíduos do modelo anterior. A ideia é que este novo modelo se concentre em corrigir os erros do seu antecessor.

**Combinação dos Modelos:** O novo modelo é adicionado ao modelo anterior, com um peso que controla sua contribuição. Este processo é repetido várias vezes, criando uma sequência de modelos que se complementam.

**Modelo Final:** O modelo final é a combinação ponderada de todos os modelos fracos. Ele tende a ser muito mais preciso do que qualquer um dos modelos individuais.

## Vantagens X Desvantagens

**Vantagens:**

possui alta precisão (Gradient Boosting é conhecido por sua alta capacidade de previsão, frequentemente superando outros algoritmos em competições de machine learning), tem flexibilidade (Pode ser usado em diversos problemas, incluindo regressão, classificação, etc.) e consegue lidar com diferentes tipos de dados (Funciona bem com dados numéricos, categóricos e mistos).

**Desvantagens:**

Alta complexidade (Pode ser mais complexo de ajustar do que outros algoritmos), Overfitting (Se não for bem ajustado, pode levar ao overfitting, onde o modelo se ajusta muito bem aos dados de treinamento, mas não sabe lidar bem com novos dados) e custo computacional (O treinamento pode ser computacionalmente intensivo, especialmente com grandes conjuntos de dados).

 ## Hiperparâmetros selecionado:

O ajuste dos hiperparâmetros é fundamental para otimizar esse modelo, por isso, os hiperparâmetros selecionados para este projeto são: 

- **learning_rate:** determina a velocidade com que o algoritmo aprende a cada iteração, valores menores indicam aprendizado mais lento, evitando overfitting. Varia entre 0 e 1, sendo o padrão 0,3
- **n_estimators:** número de árvores que serão construídas no ensemble. Este hiperparâmetro direciona a complexidade do modelo, visto que aumentar o número de n_estimators geralmente eleva a complexidade do modelo, capturando padrões mais complexos.
- **max_depth:** representa o quão profundo cada árvore pode crescer durante o treinamento do processo de boosting (número de divisões que ela terá)
- **gamma(ou min_split_loss):** É o hiperparâmetro que penaliza árvores com muitas folhas. Ele controla a decisão do algoritmo de dividir um nó em dois, calculando o ganho daquela split, como é visto na seção de matemática abaixo.
- **subsample:** controla qual porcentagem das linhas do treino é usada para fazer cada árvore nova, assim, as árvores não veem exatamente os mesmos dados 
- **colsample_bytree:** mesma ideia do subsample, mas é usado para colunas (atributos) em vez de linhas.
- **reg_lambda:** é o hiperparâmetro que penaliza pesos grandes nas folhas da árvore (valor que cada folha prevê) e os deixa menor. É a regularização L2 (Regressão Ridge).
- **reg_alpha:** é a regularização L1 (Regressão Lasso). Similar ao reg_lambda, porém nesse caso o hiperparâmetro tende a levar pesos pouco relevantes até exatamente zero, para tornar o modelo mais simples.


## Matemática por trás do XGBoost

Vamos considerar a seguinte função objetivo do XGBoost (função de perda e regularização) na iteração t que precisamos minimizar:

$$
\mathcal{L}^{(t)} = \sum_{i=1}^{n} l\Big(\underbrace{y_i}, \ \hat{y}_i^{(t-1)} + f_t(\mathbf{x}_i)\Big) + \Omega(f_t)
$$

onde o primeiro $y_i$ é o valor real conhecido do dataset de treino. Por essa ser uma função sobre a soma de várias árvores (funções),e não sobre números, ela não pode ser otimizada usando métodos de otimização tradicionais no espaço euclidiano, ou seja, é complicada demais pra resolver de forma direta. Logo, precisamos transforma-la em uma função no domínio euclidiano, para assim, usar as técnicas de otimização tradicionais.

Para alcançar-mos esse objetivo, vamos usar uma aproximação do teorema de Taylor. Nesse caso, podemos observar que a melhor aproximação linear para uma função f(x) no ponto a é:
$$
f(x) \approx f(a) + f'(a)(x-a)
$$
Onde **a** é a previsão na etapa (t-1), enquanto **(x-a)** é o novo aprendiz que precisamos adicionar na etapa (t) para minimizar o objetivo de forma gulosa.

Nessa mesma ideia, usando a aproximação de taylor para a segunda ordem, temos: 
$$
f(x) \approx f(a) + f'(a)(x-a) + \frac{1}{2}f''(a)(x-a)^2
$$

Utilizando isso em cada iteração t, podemos escrever a função objetivo (de perda) como uma função simples do novo classificador adicionado, aproximando a função para a aproximação de taylor de segunda ordem: 

$$
\mathcal{L}^{(t)} \simeq \sum_{i=1}^{n} \Big[ l(y_i, \hat{y}^{(t-1)}) + g_i f_t(\mathbf{x}_i) + \frac{1}{2} h_i f_t^2(\mathbf{x}_i) \Big] + \Omega(f_t)
$$
onde
$$
g_i = \partial_{\hat{y}^{(t-1)}} \, l(y_i, \hat{y}^{(t-1)}) \quad \text{e} \quad h_i = \partial^2_{\hat{y}^{(t-1)}} \, l(y_i, \hat{y}^{(t-1)})
$$

Por fim, ao remover as partes constantes, teremos o objetivo simplificado que será minimizado na etapa t:
$$
\tilde{\mathcal{L}}^{(t)} = \sum_{i=1}^{n} \Big[ g_i f_t(\mathbf{x}_i) + \frac{1}{2} h_i f_t^2(\mathbf{x}_i) \Big] + \Omega(f_t)
$$
O problema acima é uma soma de funções quadráticas simples de uma variável e pode ser minimizado usando técnicas de otimização conhecidas, uma vez que a parábola de segundo grau possui o ponto de minimo conhecido.

Ao longo das iterações, é preciso encontrar um algoritmo de aprendizado que alcance a máxima redução possível de perda, e por isso, precisamos medir a qualidade de uma estrutura de árvore q. Vamos usar a função de pontuação a seguir: 
$$
\tilde{\mathcal{L}}^{(t)}(q) = -\frac{1}{2} \sum_{j=1}^{T} \frac{\Big(\sum_{i \in I_j} g_i\Big)^2}{\sum_{i \in I_j} h_i + \lambda} + \gamma T
$$

Note que a “função de pontuação de qualidade” acima nos da o valor de perda mínimo para uma determinada estrutura de árvore, encontrando o peso ótimo de cada folha e substituindo de volta na função. Desse modo, a função de perda original é avaliada usando os valores de peso ideais.

Na prática, começamos com uma única raiz que tem todos os exemplos de treinamento, iteramos sobre todos os atributos e valores por atributo para avaliar cada possível redução de perda dividida (ganho = perda (instâncias pai) - ( perda (ramo esquerdo) + perda (ramo direito))).


A imagem a seguir é um exemplo da troca entre a função de perda e a de regularização, e assim, mostra o motivo pelo qual a função objetivo do XGBoost tem dois termos, já que apenas minimizar levaria ao cenário 2 (overfitting), e ignorar a perda levaria ao cenário 3 (underfitting). O equilíbrio é a meta do XGBoost, vista no cenário 4.

## Código 

### Importando o data set

Primeiro, vamos importar o data set com um conjunto de dados científico sobre supercondutores para prever a temperatura crítica de diferentes materiais, que não possui valores faltantes. Nessa tabela, temos diversos atributos, como a média da massa atomica, desvio padrão da massa atomica, entre outros. Esses atributos são aplicados para mais de 20 mil elementos. O objetivo desse modelo é acertar a temperatura crítica dos materiais (sendo esse o target).

```python
from ucimlrepo import fetch_ucirepo 
  
# fetch dataset 
superconductivty_data = fetch_ucirepo(id=464) 
  
# data (as pandas dataframes) 
X = superconductivty_data.data.features 
y = superconductivty_data.data.targets.squeeze()  # transforma DataFrame de 1 coluna em series

X
```

### Importando as bibliotecas 

Vamos importar todas as bibliotecas necessárias para o código

```python
import xgboost as xgb
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import RandomizedSearchCV
```

### Treinando o modelo 1

Agora, vamos dividir os nossos dados em treino e teste (sendo 20% para teste e 80% para treino) usando uma semente aleatória que permite reprodutibilidade. Além disso, vamos criar o modelo de regressão com os hiperparâmetros selecionados anteriormente, usando valores padrões do algoritmo.

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, 
                                                    test_size=0.2, 
                                                    random_state = 8) 
# 20% dos dados irão para os testes
# random_state serve para permitir reprodutibilidade (tipo semente no random)

modelo = xgb.XGBRegressor( # aqui criou o modelo de regressão
    n_estimators=100,     
    max_depth=6,         
    learning_rate=0.3, 
    gamma=0,
    subsample=1,
    colsample_bytree=1,
    reg_lambda=1,
    reg_alpha=1,
    random_state=8
)

# dentro do XGBRegressor temos os hiperparametros selecionados anteriormente
```

Depois disso, vamos treinar o nosso modelo e prever a raiz do erro quadratico médio (desvio padrão), o erro médio absoluto(média da distância entre o valor previsto e o valor real) e o quanto da variação da temperatura que o modelo explica.

```python
modelo.fit(X_train, y_train) # aqui o modelo estuda os dados de treino

y_pred = modelo.predict(X_test) # modelo treinado pega os dados de teste sem as respostas e chuta
                                # a resposta (previsão)

rmse = np.sqrt(mean_squared_error(y_test, y_pred))
mae = mean_absolute_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print("RMSE:", rmse)  # raiz do erro quadratico médio (em Kelvin), penaliza mais erros grandes, desvio padrao
print("MAE:", mae)   # erro médio absoluto, A média das distâncias absolutas entre os valores previstos e os reais.
print("R²:", r2)      # quanto da variação da temperatura o modelo explica, 1 é perfeito, ou o quanto ele acerta
```

Encontramos bons resultados, mas eles podem ser melhorados....

### Procurando hiperparâmetros melhores 

Na etapa anterior, o valor dos hiperparâmetros foi escolhido tentando seguir um padrão geral. Todavia, vamos observar como o modelo treina com diferentes hiperparâmetros, e assim, encontrar o melhor entre eles para criar um modelo mais completo.

```python
parametros = {
    "n_estimators": [100, 200, 300],
    "max_depth": [5, 7, 9],
    "learning_rate": [0.01, 0.05, 0.1],
    "subsample": [0.7, 0.8, 1.0],
    "colsample_bytree": [0.7, 0.8, 1.0],
    "reg_lambda": [0, 1, 5, 10],    
    "reg_alpha": [0, 1, 5, 10],      
    "gamma": [0, 0.1, 0.5, 1]          
}

busca = RandomizedSearchCV(
    xgb.XGBRegressor(random_state=8),
    param_distributions = parametros,
    n_iter=20,                     # número de combinações testadas 
    scoring="neg_root_mean_squared_error",
    cv=5,                          # validação cruzada com 5 folds (treina e testa 5 vezes as 20 combinações)
    random_state=8,
    n_jobs=-1,                      # usa todos os núcleos disponíveis
    verbose=2
)
# verbose mostra o volume de trabalho antes de começar a busca 
busca.fit(X_train, y_train)

print("Melhores hiperparâmetros:", busca.best_params_)
```

Como podemos observar, os melhores hiperparâmetros que a busca resultou foram: 

subsample: 0.8, reg_lambda: 0, reg_alpha: 5, n_estimators: 300, max_depth: 9, learning_rate: 0.05, gamma: 0, colsample_bytree: 0.8

### Treinando o modelo 2

Agora, vamos refazer todo o processo do modelo com os hiperparâmetros encontrados na busca anterior, para otimizar o processo e as previsões.

```python
modelo = xgb.XGBRegressor( # aqui criou o modelo de regressão
    n_estimators=300,     
    max_depth=9,         
    learning_rate=0.05,
    gamma=0,
    subsample=0.8,
    colsample_bytree=0.8,
    reg_lambda=0,
    reg_alpha=5,
    random_state=8
)
modelo.fit(X_train, y_train) # aqui o modelo estuda os dados de treino

y_pred = modelo.predict(X_test) # modelo treinado pega os dados de teste sem as respostas e chuta
                                # a resposta (previsão)

rmse = np.sqrt(mean_squared_error(y_test, y_pred))
mae = mean_absolute_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print("RMSE:", rmse)  
print("MAE:", mae)   
print("R²:", r2)     
```

Como podemos ver, o modelo obteve uma melhora em todos os resultados 

### Análise dos resultados 

Primeiro, vamos plotar um gráfico que mostra a temperatura critica real dos materiais e a temperatura crítica prevista pelo modelo. Além disso, ele possui uma linha da previsão perfeita em vermelho, em que pontos perto dela indicam que o modelo acertou bem aquele material, e pontos longes mostram que o modelo teve algum erro de previsão, seja subestimando a temperatura (pontos abaixo da linha) ou superestimando ela (pontos acima da linha).

```python
y_verdadeiro = y_test
y_previsao = modelo.predict(X_test)

plt.figure(figsize=(7,7))
plt.scatter(y_verdadeiro, y_previsao, alpha=0.4, s=15)

lim_min = min(y_verdadeiro.min(), y_previsao.min())
lim_max = max(y_verdadeiro.max(), y_previsao.max())
plt.plot([lim_min, lim_max], [lim_min, lim_max], color="red", linestyle="--")

plt.xlabel("Temperatura crítica real em kelvin")
plt.ylabel("Temperatura crítica prevista em kelvin")
plt.title("Previsão vs Valor Real do XGBoost Regressor")
plt.show()
```

Podemos observar que a grande parte dos pontos estão próximos da linha perfeita, o que é consistente com o $R^2$ acima de 90% visto nos resultados, e assim, apontando que o modelo teve sucesso em suas previsões. Ainda assim, é possível notar alguns pontos longes da linha perfeita, que serão investigados no próximo gráfico.

Nessa etapa, vamos fazer um **gráfico de resíduos**, para calcular o erro com sinal de cada previsão. 

```python
residuos = y_verdadeiro - y_previsao

plt.figure(figsize=(7,5))
plt.scatter(y_previsao, residuos, alpha=0.4, s=15)
plt.axhline(0, color="red", linestyle="--")
plt.xlabel("Temperatura prevista em kelvin")
plt.ylabel("Resíduo (real e previsto)")
plt.title("Gráfico de Resíduos")
plt.show()
```

Neste gráfico, temos o padrão do erro, se ele subestima, superestima ou tem viés em alguma faixa dos valores previstos. Podemos notar que a linha vermelha representa o erro zero (se o modelo não tivesse erro), e dessa forma, vemos que os pontos se distribuem homogêneamente perto dela, e apenas alguns deles estão distantes. Dessa maneira, coerente com os resultados e com o gráfico anterior.

Finalmente, vamos observar a importância de cada atributo para esse modelo, plotando um gráfico de barra com os 10 atributos mais importantes para este modelo.

```python
importancia = pd.Series(modelo.feature_importances_, index=X.columns)
melhores_features = importancia.sort_values(ascending=False).head(10)

plt.figure(figsize=(8,6))
melhores_features.sort_values().plot(kind="barh")
plt.xlabel("Importância dos atributos")
plt.title("Top 10 atributos mais importantes do XGBoost")
plt.tight_layout()
plt.show()
```

Como podemos verificar pelo gráfico, o atributo mais importante para este modelo é claramente a faixa de condutividade térmica. Diante disso, vemos que esse resultado se conecta com a **teoria da supercondutividade**, em que a base dessa teoria implica que, na observação experimental, a supercondutividade de muitos metais apresenta uma forte dependência da temperatura crítica e do campo magnético crítico em relação à massa do isótopo metálico em estudo. Isso justamente descreve o dataset escolhido para criar este modelo.

## Comparação com Floresta aleatória

Para iniciar a nossa comparação com outros modelos, vamos observar como o modelo se comporta ao usar o algoritmo de baseline.

```python
from sklearn.dummy import DummyRegressor

modelo_baseline = DummyRegressor(strategy="mean")
modelo_baseline.fit(X_train, y_train)   # treina só com treino, igual aos outros modelos
y_pred_baseline = modelo_baseline.predict(X_test)

rmse_base = np.sqrt(mean_squared_error(y_test, y_pred_baseline))
mae_base = mean_absolute_error(y_test, y_pred_baseline)
r2_base = r2_score(y_test, y_pred_baseline)

print("RMSE:", rmse_base, " MAE:", mae_base, " R²:", r2_base)
```

Como podemos avaliar, nesse caso o RMSE e o MAE são bem mais altos que os do XGBoost, mostrando que o erro de "chutar a média" é bem maior do que o erro de um modelo que aprende com os dados. Ademais, como o baseline sempre prevê a mesma coisa (média da temperatura crítica do treino) e não usa os atributos, ela não consegue explicar nenhuma variação real dos dados, e por isso, o $R^2$ esta tão próximo de zero. Esse foi o piso da comparação.

Agora, para compararmos o XGBoost com a floresta aleatória, vamos treinar um modelo de floresta aleatória com o mesmo dataset de supercondutores e alguns hiperparâmetros com os mesmos valores.

```python
from sklearn.ensemble import RandomForestRegressor 

modelo2 = RandomForestRegressor(
    n_estimators=300,     # número de árvores na floresta
    max_depth=9,          # profundidade maxima de cada arvore
    random_state=8
)

modelo2.fit(X_train, y_train)          # treina com os mesmos dados do XGBoost
y_pred2 = modelo2.predict(X_test)    

rmse2 = np.sqrt(mean_squared_error(y_test, y_pred2))
mae2 = mean_absolute_error(y_test, y_pred2)
r2_2 = r2_score(y_test, y_pred2)

print("RMSE (Random Forest):", rmse2)
print("MAE (Random Forest):", mae2)
print("R² (Random Forest):", r2_2)
```

Como podemos perceber, os resultados da previsão desse modelo foram piores do que a previsão do modelo de XGBoost.

Abaixo, temos uma comparação dos valores da floresta aleatória e do XGBoost em uma tabela, e uma discussão sobre esses resultados.

```python
comparacao = pd.DataFrame({
    "Modelo": ["XGBoost", "Random Forest"],
    "RMSE": [rmse, rmse2],
    "MAE": [mae, mae2],
    "R²": [r2, r2_2]
})
print(comparacao)
```

Como pode ser visto, o RMSE teve valores maiores que no XGBoost, mesmo com hiperparâmetros de mesmo valor. Ou seja, na média, o XGBoost erra menos Kelvin ao prever a temperatura crítica dos materiais do que a floresta aleatória. A diferença do MAE é a mais expressiva, uma vez que a floresta aleatória erra em média quase 2 K a mais por previsão (valores de 7.11 e 5.17) uma diferença de quase 40% no erro médio absoluto. A quantidade da variação da temperatura que esse modelo explica ($R^2$) também é um pouco menor em relação a do XGBoost (um pouco mais de 3%).

### Mas, por que isso acontece?

Como aprendemos ao longo desse notebook, o XGBoost corrige erros sequencialmente (o caso do boosting), e dessa maneira, cada árvore nova é feita especificamente para corrigir o erro de seus antecessores, e com isso, consegue captar padrões mais sutis nos dados. Entretanto, a floresta aleatória treina as árvores de forma independente (o bagging descrito anteriormente), e simplesmente tira a média das previsões (como se ganhasse a previsão que teve mais "votos").  Por causa disso, seus erros não são analisados e nem corrigidos, e assim, tem um modelo menos seguro. 


## Conclusão

Assim, com esse dataset grande, foi possível concluir que o XGBoost, com os seus hiperparâmetros otimizados, é uma escolha excelente de algoritmo em comparação com a floresta aleatória por exemplo. O seu alto poder de corrigir os seus erros é um dos principais motivos que esse modelo é tão forte para fazer as previsões.
